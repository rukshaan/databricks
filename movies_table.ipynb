##Column Filtering

```python
data_trimmed = df.select('title','imdb_rating','industry')
data_trimmed.show(37,truncate=False)
```

### Raw filtering

```python
df_2000_2010 = df.filter((df['release_year'] >= 2000) & (df['release_year'] <= 2010))
display(df_2000_2010)
```

```python
from pyspark.sql.functions import col
df_filtered = df.filter((col('release_year') >= 2000) & (col('release_year') <= 2010))
display(df_filtered)
```

```python
df.filter(col('release_year').between(2000,2010)).show(5)
```

```python
marvel_movies = df.filter(col('studio') == 'Marvel Studios')
display(marvel_movies)
display(marvel_movies)
df.filter(col('studio').like('%Marvel'))
```

```python
unique_industries = df.select("industry").distinct().show()
```

```python
unique_languages = df.select("language").distinct()
display(unique_languages)
```

```python
df.withColumn('Profit',col('revenue')-col('budget')).show()
```

```python
renamed_revenue = df.withColumnRenamed('revenue','total_revenue')
display(renamed_revenue)
```
