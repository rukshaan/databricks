```python
df = spark.read \
    .format('csv') \
    .option('inferSchema',True) \
    .option('header',True) \
    .option('dateFormat','dd-mm-yyyy') \
    .load("/Volumes/workspace/default/raw_data/orders.csv")
display(df)
```
