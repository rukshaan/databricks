```python
from pyspark.sql import functions as F,types as T
data =[
    ("2017-01-01",32.0,6.0,"Rainy"),
    ("2017-01-02",None,9.0,"Sunny"),
    ("2017-01-03",28.0,None,"Snowy"),
    ("2017-01-04",23.0,0.5,None),
    ("2017-01-05",32.0,None,"Sunny"),
    ("2017-01-06",None,None,"Rainy"),
    ("2017-01-07",None,None,None),
    ("2017-01-08",32.0,7.2,"Sunny"),
    ("2017-01-09",31.0,9.0,"Sunny")
]
schema = "day string,tempreature double,windspeed double,event string"
df = spark.createDataFrame(data,schema)
df = df.withColumn("day",F.to_date(F.col("day"),"yyyy-MM-dd"))
display(df)
```

```python
df.createOrReplaceTempView('Weather') #session
```

```python
%sql
SELECT 
    event,
    ROUND(AVG(tempreature), 1) AS avg_tempreature
FROM weather
GROUP BY event
ORDER BY avg_tempreature DESC;
```
